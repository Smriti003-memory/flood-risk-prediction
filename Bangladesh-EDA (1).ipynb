3. Data Visualization

```python
correlation=df.corr()
correlation
```

```python
# Visualisation of the corralation table


plt.figure(figsize=(12,8), dpi=77)
sns.heatmap(correlation, linecolor='white',linewidths=0.1, annot=True)
plt.title('Correlation Matrix'.upper(), size=19, pad=13)
plt.xlabel('Flood Data')
plt.ylabel('Flood Data')
plt.xticks(rotation=33)
plt.show()
```

```python
#pie chart

class_0 = (df['grid_code'] == 0).sum()
class_1 = (df['grid_code'] == 1).sum()
proportions = [class_0, class_1]
print(proportions)
print("Show")
plt.figure(figsize=(12, 8), dpi=77)
plt.pie(
    proportions, 
    labels=['Class 0 (Absence)', 'Class 1 (Presence)'], 
    explode=(0.05, 0), 
    startangle=90, 
    autopct='%1.1f%%', 
    shadow=False
)
plt.axis('equal')
plt.title("Grid Code Proportion", size=17, pad=13)
plt.show()
```

```python
#bar plot
# DEM vs grid

plt.figure(figsize=(10,8), dpi=77)
sns.barplot(x="grid_code", y="DEM", data=df)
plt.title("Elevation (DEM) by Grid Code", size=17, pad=13)
plt.show()
```

```python
#countplot
plt.figure(figsize=(10, 8), dpi=77)
sns.countplot(x=df['grid_code'], hue=df['grid_code'])
plt.title("Comparison: Grid Code Distribution", size=17, pad=13)
plt.legend(['Class 0 (Absence)', 'Class 1 (Presence)'], loc='upper right', prop={'size': 12})
plt.show()
```

```python
#box pot
#grid_code vs DEM

plt.figure(figsize=(10,8), dpi=77)
sns.boxplot(x="grid_code", y="DEM", data=df)
plt.title("Comparison: DEM (Elevation) of Class 0 / Class 1", size=17, pad=13)
plt.ylabel('DEM')
plt.xlabel(' ')
plt.xticks([0, 1], ['Class 0 (Absence)', 'Class 1 (Presence)'])
plt.show()
```

```python
# Histogram: DEM distribution
plt.figure(figsize=(8, 6))
plt.hist(df['DEM'], bins=20, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('DEM Distribution')
plt.xlabel('DEM')
plt.ylabel('Frequency')
plt.show()
```

```python
# Scatter plot: DEM vs NDVI

plt.figure(figsize=(8, 6))
plt.scatter(df['DEM'], df['NDVI'], c=df['grid_code'], cmap='coolwarm', alpha=0.6)
plt.title('DEM vs NDVI')
plt.xlabel('DEM')
plt.ylabel('NDVI')
plt.colorbar(label='grid_code')
plt.show()
```

```python
#line plot:DEM vs NDVI

plt.figure(figsize=(10, 6), dpi=77)
sns.lineplot(x=df['DEM'].round(), y=df['NDVI'], hue=df['grid_code'], palette='Set1', marker='o')
plt.title("Trend: NDVI across Elevation (DEM)", size=17, pad=13)
plt.xlabel("DEM (Elevation)", size=12)
plt.ylabel("NDVI", size=12)
plt.legend(['Class 0 (Absence)', 'Class 1 (Presence)'], loc='upper left', prop={'size': 11})
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
x=df.drop(columns=['grid_code'])
y=df['grid_code']
```

```python
reg = LogisticRegression()
reg.fit(x, y)
```

```python
print(reg.coef_)

print(reg.intercept_)
```

```python
y_pred= reg.predict(x)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score
y_test=df['grid_code']
accuracy_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt
from sklearn import metrics
confusion_matrix = metrics.confusion_matrix(y_test,y_pred)
cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix = confusion_matrix, display_labels = [False, True])
cm_display.plot()
plt.show()
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

```
